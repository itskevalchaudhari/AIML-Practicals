```python
# Practical 1: Identification of Association Between
# Dependent and Independent Variables

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries imported successfully!")
```

```python
from sklearn.datasets import load_diabetes

# Load the Diabetes dataset
diabetes = load_diabetes()

# Create a DataFrame
df = pd.DataFrame(diabetes.data, columns=diabetes.feature_names)
df['target'] = diabetes.target

# Display the first five records
df.head()
```

```python
# Display dataset information
print("Dataset Shape:", df.shape)
print("\nIndependent Variables:")
print(diabetes.feature_names)

print("\nDependent Variable:")
print("target")

print("\nDataset Information:")
df.info()
```

```python
# Calculate correlation of all variables with the dependent variable
correlation = df.corr()['target'].sort_values(ascending=False)

print("Correlation of Independent Variables with Target:\n")
print(correlation)
```

```python
# Correlation of independent variables with the dependent variable
independent_correlation = correlation.drop('target')

print("Association of Independent Variables with Dependent Variable:\n")
print(independent_correlation)
```

```python
# Create a correlation matrix
correlation_matrix = df.corr()

# Plot the correlation heatmap
plt.figure(figsize=(12, 8))
sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f',
    linewidths=0.5
)

plt.title('Correlation Heatmap of Independent and Dependent Variables')
plt.tight_layout()
plt.show()
```

```python
# Scatter plot for the strongest positively associated variable
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df,
    x='bmi',
    y='target'
)

plt.title('Association Between BMI and Target')
plt.xlabel('BMI')
plt.ylabel('Target')
plt.tight_layout()
plt.show()
```

# Result and Conclusion

The association between the independent variables and the dependent variable was identified using Pearson correlation analysis.

The correlation analysis shows that **BMI (0.586)** has the strongest positive association with the target variable, followed by **s5 (0.566)** and **bp (0.441)**. The variable **s3 (-0.395)** shows a moderate negative association with the target, while **sex (0.043)** has a very weak association.

The correlation heatmap provides a visual representation of the relationships among the variables. The scatter plot between BMI and the target also shows a positive trend.

Therefore, the dependent and independent variables were successfully identified and their associations were analyzed using correlation analysis.
