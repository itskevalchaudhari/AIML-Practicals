## Theory

Object detection is a computer vision technique used to identify and locate objects in an image or video. Unlike image classification, object detection provides both the **class label** of an object and its **location** using a bounding box.

CCTV footage consists of a sequence of video frames. Object detection can be applied to each frame to identify objects such as people, cars, bicycles, buses, and other objects.

### YOLO (You Only Look Once)

YOLO is a real-time object detection algorithm that performs object detection in a single neural network evaluation. It processes the complete image at once and predicts bounding boxes and class probabilities simultaneously.

In this practical, a **pre-trained YOLO model** is used to detect objects in CCTV footage.

### Main Components of YOLO

1. **Bounding Box** – Represents the location of a detected object.
2. **Class Label** – Identifies the detected object.
3. **Confidence Score** – Indicates how confident the model is about the detection.
4. **Non-Maximum Suppression (NMS)** – Removes overlapping duplicate bounding boxes.

### Bounding Box Parameters

A detected bounding box can be represented using:

- x = center x-coordinate
- y = center y-coordinate
- w = width of the bounding box
- h = height of the bounding box

The bounding box is commonly represented as:

**(x, y, w, h)**

### Intersection over Union (IoU)

IoU measures the overlap between the predicted bounding box and the actual bounding box.

**IoU = Area of Intersection / Area of Union**

A higher IoU indicates better overlap between the two bounding boxes.

### Confidence Score

The confidence score represents the model's confidence that a detected bounding box contains an object of a particular class.

Object detection can be used in CCTV surveillance for applications such as people counting, vehicle detection, traffic monitoring, security monitoring, and suspicious activity analysis.

```python
# Import required libraries

from ultralytics import YOLO
import cv2
import matplotlib.pyplot as plt

print("Ultralytics YOLO imported successfully.")
```

```python
# Load the pre-trained YOLO model

model = YOLO("yolo11n.pt")

print("YOLO model loaded successfully.")
```

```python
# Upload CCTV video

from google.colab import files

uploaded = files.upload()
```

```python
# Perform object detection on CCTV footage

input_video = "Felony_Hit-and-Run.webm"

results = model.predict(
    source=input_video,
    conf=0.25,
    save=True
)

print("Object detection completed successfully.")
```

```python
# Find the processed CCTV video

import os
import glob

video_files = []

for root, dirs, files in os.walk("runs"):
    for file in files:
        if file.lower().endswith((".mp4", ".avi", ".mov", ".webm", ".mkv")):
            video_files.append(os.path.join(root, file))

print("Processed video files found:")

for file in video_files:
    print(file)
```

```python
# Display the processed CCTV video

from IPython.display import Video, display

output_video = "runs/detect/predict/Felony_Hit-and-Run.avi"

print("Processed video:", output_video)

display(Video(output_video, embed=True))
```

```python
# Convert the detected AVI video to MP4

import subprocess

input_video = "runs/detect/predict/Felony_Hit-and-Run.avi"
output_video = "runs/detect/predict/Felony_Hit-and-Run_Detected.mp4"

subprocess.run([
    "ffmpeg",
    "-y",
    "-i", input_video,
    "-c:v", "libx264",
    "-pix_fmt", "yuv420p",
    output_video
], check=True)

print("Detected video converted to MP4 successfully.")
print("Output:", output_video)
```

```python
# Display the final detected CCTV video

from IPython.display import Video, display

final_video = "runs/detect/predict/Felony_Hit-and-Run_Detected.mp4"

print("Final detected video:")
print(final_video)

display(Video(final_video, embed=True))
```

```python
# Object Detection Summary

from collections import Counter

detected_objects = []

for result in results:
    if result.boxes is not None:
        for cls in result.boxes.cls:
            detected_objects.append(model.names[int(cls)])

object_counts = Counter(detected_objects)

print("Object Detection Summary")
print("------------------------")

for obj, count in object_counts.items():
    print(f"{obj}: {count} detections")

print("\nTotal detections:", len(detected_objects))
```
